```python
document = [] 
# document to hold all the data that need to upload in the  vectore database
```

```python
from langchain_community.vectorstores import FAISS
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.embeddings import OllamaEmbeddings
```

#### upload a text file

```python
path = "" # write the relative path of text file here
from langchain_community.document_loaders import TextLoader  
text_loader = TextLoader(path)
document.extend(text_loader.load())
```

#### upload a PDF 

```python
path = "" # write the relative path of the pdf here
from langchain_community.document_loaders import PyPDFLoader
pdf_loader = PyPDFLoader("attention.pdf")
document.extend(pdf_loader.load())
```

#### Get the data from a Website

```python

from langchain_community.document_loaders import WebBaseLoader
# web_loader = WebBaseLoader(web_path="https://en.wikipedia.org/wiki/Cardamom")  # to load single web page
web_loader = WebBaseLoader(web_paths=("https://en.wikipedia.org/wiki/Cardamom" ,"https://jntuh.org/home" ))  
# to load multiple web pages
document.extend(web_loader.load())
```

## Splitting the document data

splitting is a use process where we split the data into the chunks, which makes it easy to read and store in database

```python
splitter = RecursiveCharacterTextSplitter(chunk_size = 1000 , chunk_overlap = 300)
splitted_document = splitter.split_documents(document)
```

## Embedding and Vector DataBase

Embedding is used to convert the strings into the vectors. And Vector DataBase is used to store these vectors

```python
# Note :- to run this line, we need to download the gemma:2b model using ollama in the system, 
# you can also use other embedding models also from other platforms like OpenAI and HuggingFace.
my_embeddings = OllamaEmbeddings(model="gemma:2b")
```

```python
vector_data_base = FAISS.from_documents(splitted_document,my_embeddings)
```
